# Investigating XEQT

This notebook recomputes the report numbers from Yahoo Finance weekly prices in CAD.
Run the download script first, or run the cells in order. Data are in data/.

```
import pandas as pd, numpy as np, json
rets=pd.read_csv('data/weekly_returns.csv', index_col=0, parse_dates=True)
print(rets.shape)
print(rets.head())
```

```
# Annual return, volatility, Sharpe with a stated 2% risk free rate, max drawdown, weekly CVaR
for c in rets.columns:
    s=rets[c]; ann=s.mean()*52; vol=s.std()*np.sqrt(52)
    growth=(1+s).cumprod(); dd=growth/growth.cummax()-1
    print(c, round(ann*100,2), round(vol*100,2), round((ann-0.02)/vol,2), round(dd.min()*100,2), round(s[s<=s.quantile(0.05)].mean()*100,2))
```

```
# Calendar year total returns
for y in [2020,2022,2023,2024,2025]:
    x=rets[str(y)]; print(y, {c: round(((1+x[c]).prod()-1)*100,2) for c in x.columns})
```
